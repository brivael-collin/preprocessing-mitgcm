Converts the binary files between big-endian (the format used with MITgcm) and little-endian, for a computer that would need little-endian files.
This Notebook allows for a swift conversion of the binary files.

```python
import numpy as np
import glob
import os

from lib.config import ConfigObject
```

```python
# Convert in place the binary files of the current simulation (in 2-preprocessing-outputs)
config = ConfigObject('../../config.json')
input_folder = os.path.join(config.Paths.PreprocessingOutputs, config.SimulationName, 'binary_data')
output_folder = input_folder

input_datatype = '>f8'
output_datatype = '<f8'
```

```python
input_files = glob.glob(os.path.join(input_folder, '*.bin'))
```

```python
for file in input_files:
    with open(file, 'rb') as fid:
        binary_data = np.fromfile(fid, dtype=input_datatype)
    
    # Convert
    data = binary_data.astype(output_datatype)
    # Write to file
    fid = open(os.path.join(output_folder, os.path.basename(file)), 'wb')
    data.tofile(fid)
    fid.close()
```
